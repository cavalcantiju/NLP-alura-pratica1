```python
import pandas as pd
import nltk
from nltk import tokenize
from wordcloud import WordCloud

df = pd.read_csv('/content/am_scrape_final.csv')

df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

todas_palavras = ' '.join([texto for texto in df.Review])
token_espaco = tokenize.WhitespaceTokenizer()
token_frase = token_espaco.tokenize(todas_palavras)
frequencia = nltk.FreqDist(token_frase)
df_frequencia = pd.DataFrame({'Palavra': list(frequencia.keys()),
                              'Frequência': list(frequencia.values())})
df_frequencia.head() #freq da primeira rase
plt.figure(figsize=(20,7))
ax = sns.barplot(data=df_frequencia.nlargest(columns='Frequência', n=20), x='Palavra', y='Frequência', color='gray')
ax.set(ylabel='Contagem')
plt.show()
```
